***3) Exploratory Data Analysis (EDA) and Visualization***

**Campaign Performance:**

```python
# Scatter Plot – Acquisition Cost vs ROI
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Acquisition_Cost',
    y='ROI',
    hue='Campaign_Type',
    palette='Set1'
)

plt.title("Relationship Between Acquisition Cost and ROI")
plt.xlabel("Acquisition Cost")
plt.ylabel("ROI")
plt.legend(bbox_to_anchor=(1, 1))

plt.tight_layout()
plt.show()
```

```python
# Bar Chart – Average Conversion Rate by Channel and Campaign Type
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='Channel_Used',
    y='Conversion_Rate',
    hue='Campaign_Type',
    palette='Set2'
)

plt.title("Average Conversion Rate by Channel and Campaign Type")
plt.xlabel("Channel Used")
plt.ylabel("Average Conversion Rate")
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
# Box Plot – Engagement Score by Campaign Type
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x='Campaign_Type',
    y='Engagement_Score',
    hue='Campaign_Type',
    palette='Set1',
    legend=False
)

plt.title("Distribution of Engagement Score by Campaign Type")
plt.xlabel("Campaign Type")
plt.ylabel("Engagement Score")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Bar Chart – Average ROI by Company
avg_roi = df.groupby('Company')['ROI'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_roi.index,
    y=avg_roi.values,
    hue=avg_roi.index,
    palette='Set2',
    legend=False
)

plt.title("Average ROI by Company")
plt.xlabel("Company")
plt.ylabel("Average ROI")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
 # Heatmap – Engagement Score and Conversion Rate
corr_matrix = df[['Engagement_Score', 'Conversion_Rate']].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f'
)

plt.title("Correlation Between Engagement Score and Conversion Rate")
plt.xticks(rotation=20)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

**Customer Segmentation:**

```python
# Count Plot – Target Audience
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='Target_Audience',
    hue='Target_Audience',
    palette='Set1',
    legend=False
)

plt.title("Distribution of Target Audience")
plt.xlabel("Target Audience")
plt.ylabel("Number of Campaigns")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Bar Chart - Highest Conversion Rate Customer Segment for Each Language
segment_conversion = (
    df.groupby(['Language', 'Customer_Segment'])['Conversion_Rate']
      .mean()
      .reset_index()
)

highest_segment = segment_conversion.loc[
    segment_conversion.groupby('Language')['Conversion_Rate'].idxmax()
]

plt.figure(figsize=(10, 6))

sns.barplot(
    data=highest_segment,
    x='Language',
    y='Conversion_Rate',
    hue='Customer_Segment',
    palette='Set1'
)

plt.title("Highest Conversion Rate Customer Segment by Language")
plt.xlabel("Language")
plt.ylabel("Average Conversion Rate")

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
# Box Plot – Acquisition Cost by Customer Segment and Channel
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='Customer_Segment',
    y='Acquisition_Cost',
    hue='Channel_Used',
    palette='Set2'
)

plt.title("Acquisition Cost by Customer Segment and Channel")
plt.xlabel("Customer Segment")
plt.ylabel("Acquisition Cost")
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
# Bar Chart – Average Conversion Rate by Language
avg_conversion_language = (
    df.groupby('Language')['Conversion_Rate']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=avg_conversion_language.index,
    y=avg_conversion_language.values,
    hue=avg_conversion_language.index,
    palette='viridis',
    legend=False
)

plt.title("Average Conversion Rate by Language")
plt.xlabel("Language")
plt.ylabel("Average Conversion Rate")

plt.tight_layout()
plt.show()
```

**Channel Effectiveness:**

```python
# Bar Chart – Engagement Score by Channel and Campaign Type
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='Channel_Used',
    y='Engagement_Score',
    hue='Campaign_Type',
    palette='Set3'
)

plt.title("Engagement Score by Channel and Campaign Type")
plt.xlabel("Channel Used")
plt.ylabel("Average Engagement Score")
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
# Pie Chart – Total ROI by Channel
channel_roi = df.groupby('Channel_Used')['ROI'].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    channel_roi.values,
    labels=channel_roi.index,
    colors = ['Green', 'Red','Orange','Blue', 'Purple', 'Pink'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Distribution of Total ROI by Channel")

plt.show()
```

```python
# Scatter Plot – Clicks vs Impressions by Channel
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Clicks',
    y='Impressions',
    hue='Channel_Used',
    palette='Set3'
)

plt.title("Relationship Between Clicks and Impressions by Channel")
plt.xlabel("Clicks")
plt.ylabel("Impressions")

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

**Time-Based Analysis:**

```python
# Histogram

plt.figure(figsize=(10, 6))

plt.hist(df['Duration'],
         bins=20,
         color='yellow',
         edgecolor='red')

plt.title("Distribution of Trip Duration")
plt.xlabel("Duration")
plt.ylabel("Number of Trips")

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# Line Chart – Conversion Rate Over Date for Each Company
conversion_date_company = (
    df.groupby(['Date', 'Company'])['Conversion_Rate']
      .mean()
      .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=conversion_date_company,
    x='Date',
    y='Conversion_Rate',
    hue='Company'
)

plt.title("Conversion Rate Over Time by Company")
plt.xlabel("Date")
plt.ylabel("Average Conversion Rate")

plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1, 1))

plt.tight_layout()
plt.show()
```

```python
# line chart based on month Conversion Rate Over Date for Each Company
# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Create monthly date
df['Month'] = df['Date'].dt.to_period('M').dt.to_timestamp()

# Calculate monthly average Conversion Rate for each Company
conversion_month_company = (
    df.groupby(['Month', 'Company'])['Conversion_Rate']
      .mean()
      .reset_index()
)

# Line chart
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=conversion_month_company,
    x='Month',
    y='Conversion_Rate',
    hue='Company',
    marker='o'
)

plt.title("Monthly Conversion Rate by Company")
plt.xlabel("Month")
plt.ylabel("Average Conversion Rate")

plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1, 1))

plt.tight_layout()
plt.show()
```

```python
# Line Chart – Engagement Score Over Date
engagement_date = (
    df.groupby('Date')['Engagement_Score']
      .mean()
      .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=engagement_date,
    x='Date',
    y='Engagement_Score'
)

plt.title("Engagement Score Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Average Engagement Score")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**Geographic Analysis:**

```python
# Bar Chart – Acquisition Cost by Location
location_cost = (
    df.groupby('Location')['Acquisition_Cost']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=location_cost.index,
    y=location_cost.values,
    hue=location_cost.index,
    palette='Set1',
    legend=False
)

plt.title("Average Acquisition Cost by Location")
plt.xlabel("Location")
plt.ylabel("Average Acquisition Cost")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Bar Chart – Conversion Rate by Location and Target Audience
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Location',
    y='Conversion_Rate',
    hue='Target_Audience',
    palette='Set2'
)

plt.title("Conversion Rate by Location and Target Audience")
plt.xlabel("Location")
plt.ylabel("Average Conversion Rate")
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
# Pie Chart – Proportion of ROI by Location
location_roi = df.groupby('Location')['ROI'].sum()

plt.figure(figsize=(8, 8))

plt.pie(
    location_roi.values,
    labels=location_roi.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Proportion of ROI by Location")

plt.show()
```
